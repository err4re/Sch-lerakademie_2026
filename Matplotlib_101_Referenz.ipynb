## 0. Warum matplotlib?

- **Standardwerkzeug** zum Plotten in Python, fast jede wissenschaftliche Grafik in Python entsteht damit.
- Arbeitet direkt mit **NumPy-Arrays** zusammen: ein Array `x` und ein gleich langes Array `y` reichen für einen Plot.
- Sehr **feingranular einstellbar**: Titel, Achsen, Farben, Marker, Skalierung, fast alles lässt sich per Argument ändern.
- Die Basis-Funktionalität heißt `matplotlib.pyplot`, üblich importiert als `plt`.

```python
import numpy as np
import matplotlib.pyplot as plt
```

---
## 1. Der einfache Linienplot: `plt.plot`

`plt.plot(x, y)` zeichnet zu jedem `x`-Wert den zugehörigen `y`-Wert als Punkt und verbindet die Punkte mit einer Linie. `x` und `y` müssen **gleich lang** sein, meist NumPy-Arrays.

`plt.show()` zeigt den fertigen Plot an.

```python
x = np.linspace(0, 10, 50)   # 50 x-Werte zwischen 0 und 10
y = np.sin(x)                 # dazugehoerige y-Werte

plt.plot(x, y)
plt.show()
```

---
## 2. Titel und Achsenbeschriftung

- `plt.title("...")`: Überschrift über dem Plot
- `plt.xlabel("...")`, `plt.ylabel("...")`: Beschriftung der Achsen

Diese Befehle stehen **nach** `plt.plot(...)` und **vor** `plt.show()`, in derselben Zelle.

```python
plt.plot(x, y)
plt.title("Sinusfunktion")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

## 3. Streudiagramm: `plt.scatter`

`plt.scatter(x, y)` zeichnet die Punkte **ohne** sie zu verbinden. Das eignet sich für einzelne Messwerte, bei denen eine Verbindungslinie keinen Sinn ergibt (im Gegensatz zu einer glatten Funktion wie oben).

```python
messreihe_x = np.array([1, 2, 3, 4, 5])
messreihe_y = np.array([2.1, 3.9, 6.2, 7.8, 10.1])

plt.scatter(messreihe_x, messreihe_y)
plt.title("Messwerte")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

---
## 4. Marker und Linienstil ändern

Sowohl `plt.plot` als auch `plt.scatter` akzeptieren zusätzliche Argumente:

| Argument | Wirkung | Beispiele |
|---|---|---|
| `color` | Farbe | `"red"`, `"g"`, `"darkorange"` |
| `marker` | Symbol pro Punkt | `"o"`, `"x"`, `"s"`, `"^"`, `"d"` |
| `linestyle` (nur `plot`) | Linienart | `"-"`, `"--"`, `":"`, `"-."` |
| `markersize` / `s` | Punktgröße (`plot` bzw. `scatter`) | Zahl, z. B. `8` |
| `label` | Name für die Legende | `"Messung 1"` |

Mit `label="..."` gibst du einer Kurve einen Namen; `plt.legend()` zeigt dann eine Legende an.

```python
plt.plot(x, y, color="darkorange", linestyle="--", marker="o", markersize=4, label="sin(x)")
plt.title("Marker und Linienstil")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
plt.scatter(messreihe_x, messreihe_y, marker="^", color="crimson", s=80)   # s = Punktgroesse
plt.title("Messwerte mit anderem Marker")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

## 5. Achsenskalierung ändern

- `plt.xlim(min, max)`, `plt.ylim(min, max)`: nur einen bestimmten Ausschnitt der Achse anzeigen
- `plt.xscale("log")`, `plt.yscale("log")`: Achse logarithmisch statt linear darstellen

Logarithmische Achsen sind nützlich, wenn Werte über mehrere Größenordnungen streuen (z. B. 1 bis 1000).

```python
x_pos = np.linspace(1, 1000, 200)
y_pos = x_pos**2

plt.plot(x_pos, y_pos)
plt.title("Standard-Achsen (linear)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
plt.plot(x_pos, y_pos)
plt.title("Logarithmische y-Achse")
plt.xlabel("x")
plt.ylabel("y")
plt.yscale("log")
plt.show()
```

```python
plt.plot(x_pos, y_pos)
plt.title("Ausschnitt der x-Achse")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 100)   # nur den Bereich 0 bis 100 anzeigen
plt.show()
```

## 6. Arrays und Funktionen zum Plotten

Statt Werte von Hand einzutippen, definiert man meist eine **Funktion** und wertet sie über ein mit `np.linspace` erzeugtes Array aus. Vorteil: dieselbe Funktion lässt sich mit unterschiedlichen Parametern mehrfach plotten, z. B. um Parameter zu vergleichen.

```python
def gedaempfte_schwingung(t, A=1.0, tau=2.0, omega=5.0):
    return A * np.exp(-t / tau) * np.cos(omega * t)

t = np.linspace(0, 10, 300)

plt.plot(t, gedaempfte_schwingung(t), label="A=1.0, tau=2.0")
plt.plot(t, gedaempfte_schwingung(t, A=1.5, tau=4.0), label="A=1.5, tau=4.0")
plt.title("Gedaempfte Schwingung fuer verschiedene Parameter")
plt.xlabel("t")
plt.ylabel("Auslenkung")
plt.legend()
plt.show()
```
