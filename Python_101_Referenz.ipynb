## 1. Das Jupyter Notebook

### 1.1 Was ist eine Zelle?
Ein Notebook besteht aus **Zellen**. Eine Zelle ist die kleinste Einheit, die man ausführen kann. Es gibt zwei wichtige Typen:

| Typ | Inhalt | Zweck |
|---|---|---|
| **Code** | Python-Code | wird ausgeführt, erzeugt Output |
| **Markdown** | Text | Erklärungen, Formeln, Überschriften |

### 1.2 Was steht links von der Zelle?
Links von jeder **Code**-Zelle steht eine Markierung in eckigen Klammern:

- `[ ]` — Zelle wurde noch nicht ausgeführt
- `[*]` — Zelle läuft gerade
- `[7]` — Zelle war die **7. ausgeführte Zelle** seit dem letzten Kernel-Start

Die Zahl ist also **kein** Zeilennummer und **keine** Position im Notebook, sondern ein *Ausführungszähler*. Markdown-Zellen haben keine solche Markierung.

### 1.3 Zellen hinzufügen
- Auf das `+`-Symbol in der Werkzeugleiste klicken, **oder**
- `Esc` drücken (blauer Rand = Kommando-Modus), dann:
  - `b` — neue Zelle **b**elow (darunter)
  - `a` — neue Zelle **a**bove (darüber)
  - `m` — Zelle in **M**arkdown umwandeln
  - `y` — Zelle in Code umwandeln
  - `d`, `d` — Zelle löschen
- `Enter` bringt dich zurück in den Edit-Modus (grüner Rand).

### 1.4 Was ist ein Kernel?
Der **Kernel** ist der Python-Prozess im Hintergrund, der deinen Code ausführt und **alle Variablen im Speicher hält**. Das Notebook ist nur die Oberfläche, der Kernel ist das Gedächtnis.

- **Unterbrechen:** Werkzeugleiste `■` (Stop) oder Menü *Kernel → Interrupt Kernel*. Nutzt du bei Endlosschleifen. Variablen bleiben erhalten.
- **Neu starten:** Menü *Kernel → Restart Kernel*. Der Speicher wird **komplett geleert**, alle Variablen sind weg.
- *Kernel → Restart Kernel and Run All Cells* ist der beste Test, ob dein Notebook wirklich von oben nach unten funktioniert.

### 1.5 Eine Idee pro Zelle, und die Reihenfolge zählt

**Faustregel: eine Idee / eine Aufgabe pro Zelle.** Kleine Zellen kann man einzeln testen und wiederholen, ohne alles neu zu rechnen.

**Achtung die wichtigste Falle:** Ein Notebook wird in der Reihenfolge ausgeführt, in der *du* die Zellen anklickst, **nicht** in der Reihenfolge, in der sie untereinander stehen. Die Ausführung ist also **nicht zwingend linear**.

Das heißt: Ein Notebook kann bei dir funktionieren und trotzdem kaputt sein, weil du eine Zelle zwischendurch geändert oder mehrfach ausgeführt hast. Kontrolliere die Zahlen in `[ ]`, laufen sie von oben nach unten aufsteigend durch?

````python
x = 10
print("x ist jetzt", x)
````

````python
# Fuehre diese Zelle mehrfach hintereinander aus und beobachte das Ergebnis.
x = x + 5
print("x ist jetzt", x)
````

### 1.6 Was ist der Output einer Zelle?

Unter jeder Code-Zelle erscheint ihr **Output**. Er entsteht auf zwei Wegen:

1. **Automatisch:** Der Wert des **letzten Ausdrucks** einer Zelle wird angezeigt (mit `Out[n]:` markiert), nur der letzte!
2. **Explizit:** Alles, was `print()` ausgibt, erscheint (beliebig oft, ohne `Out[n]:`).

Zuweisungen (`a = 5`) erzeugen keinen Output.

````python
a = 3
b = 4
a + b      # letzter Ausdruck -> wird automatisch angezeigt
````

````python
a + b      # dieser Wert verschwindet
a * b      # nur dieser wird angezeigt
````

````python
print(a + b)   # beide erscheinen, weil explizit ausgegeben
print(a * b)
````

### 1.7 Markdown-Zellen

**Markdown** ist eine einfache Auszeichnungssprache: man schreibt normalen Text und markiert Formatierung mit wenigen Sonderzeichen. Beim Ausführen (`Shift + Enter`) wird der Text sauber gesetzt.

## 2. Variablen und Datentypen

Eine **Variable** ist ein Name, der auf einen Wert im Speicher zeigt. Zuweisung mit `=`:

```python
name = wert
```

Python fragt den Typ nicht ab — er ergibt sich aus dem Wert. Mit `type()` kannst du ihn prüfen.

````python
n = 42          # int   = ganze Zahl
x = 3.14        # float = Gleitkommazahl
print(n, type(n))
print(x, type(x))
````

````python
# Vorsicht: int-Division und float-Division
print(7 / 2)     # 3.5  -> immer float
print(7 // 2)    # 3    -> ganzzahlige Division
print(7 % 2)     # 1    -> Rest
print(2 ** 10)   # 1024 -> Potenz
````

### 2.1 Liste (`list`)
Eine geordnete, veränderbare Sammlung beliebiger Objekte. **Indizierung beginnt bei 0.**

````python
werte = [10, 20, 30, 40, 50]

print(werte[0])     # erster Eintrag
print(werte[2])     # dritter Eintrag
print(werte[-1])    # LETZTER Eintrag  (negative Indizes zaehlen von hinten)
print(werte[-2])    # vorletzter Eintrag
print(len(werte))   # Anzahl der Eintraege
````

````python
# Ausschnitte (Slicing): [start:stop]  -- stop ist NICHT enthalten
print(werte[1:4])    # [20, 30, 40]
print(werte[:3])     # ab Anfang
print(werte[2:])     # bis Ende

# Listen sind veraenderbar
werte[0] = 99
werte.append(60)     # haengt hinten an
print(werte)
````

### 2.2 NumPy-Array (`np.array`)
Für Zahlen ist eine Liste unpraktisch: Rechnen geht nur elementweise per Schleife. Ein **NumPy-Array** ist ein Block gleichartiger Zahlen im Speicher — dadurch schnell und mathematisch direkt verwendbar.

````python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[0])      # erster Eintrag  -- gleiche Indizierung wie bei Listen
print(a[-1])     # letzter Eintrag
print(a[1:4])    # Slicing funktioniert ebenso
print(a.shape)   # Form des Arrays
print(a.dtype)   # Datentyp der Eintraege
````

````python
# Der entscheidende Unterschied: Rechnen ohne Schleife (Vektorisierung)
print(a * 2)        # jedes Element mal 2
print(a + a)        # elementweise Addition
print(np.sqrt(a))   # Funktion auf alle Elemente

# Zum Vergleich: bei einer Liste bedeutet * 2 etwas voellig anderes!
print([10, 20, 30] * 2)   # Liste wird ANGEHAENGT, nicht multipliziert
````

````python
# Nuetzliche Erzeuger
print(np.zeros(5))
print(np.arange(0, 10, 2))        # von 0 bis <10 in Schritten von 2
print(np.linspace(0, 1, 5))       # 5 Werte gleichmaessig von 0 bis 1
````

## 3. `print()`

`print()` schreibt Werte in den Output einer Zelle. Es ist dein wichtigstes Werkzeug zur Fehlersuche: *Wenn du nicht weißt, was in einer Variablen steht — gib sie aus.*

Anders als der automatische Output funktioniert `print()` überall: in Schleifen, in Funktionen, mehrfach pro Zelle.

````python
print("Hallo")                    # Text
print(a)                          # Variable
print("Wert:", 42, "Einheit: m")  # mehrere Argumente, durch Leerzeichen getrennt
````

````python
# f-String: Variablen direkt in den Text einsetzen (empfohlen)
laenge = 3.14159
print(f"Die Laenge betraegt {laenge} m")
print(f"Gerundet: {laenge:.2f} m")        # 2 Nachkommastellen
print(f"Wissenschaftlich: {laenge:.3e} m")
````

## 4. Funktionen

Eine **Funktion** ist ein benannter, wiederverwendbarer Codeblock. Definition mit `def`:

```python
def name(argument1, argument2):
    ...
    return ergebnis
```

- **Argumente (Parameter)** sind die Eingaben. Übergeben kannst du alles: Zahlen, Strings, Listen, Arrays, sogar andere Funktionen.
- `return` gibt einen Wert zurück und **beendet** die Funktion sofort. Ohne `return` liefert die Funktion `None`.
- Der Funktionskörper wird **eingerückt** (4 Leerzeichen).

````python
def flaeche_kreis(r):
    A = np.pi * r**2
    return A

print(flaeche_kreis(1.0))
print(flaeche_kreis(2.5))
````

````python
# Mehrere Argumente, Standardwert und mehrere Rueckgabewerte
def fallstrecke(t, g=9.81):
    # Berechnet Fallstrecke und Endgeschwindigkeit nach Zeit t
    s = 0.5 * g * t**2
    v = g * t
    return s, v          # gibt ein Tupel zurueck

s, v = fallstrecke(0.85)                 # g nutzt den Standardwert
print(f"s = {s:.3f} m, v = {v:.3f} m/s")

s_mond, v_mond = fallstrecke(0.85, g=1.62)   # Argument per Name uebergeben
print(f"Mond: s = {s_mond:.3f} m")
````

````python
# Funktionen wirken auch auf ganze Arrays, wenn sie vektorisiert geschrieben sind
zeiten = np.linspace(0, 0.85, 5)
print(zeiten)
print(fallstrecke(zeiten)[0])
````

### Warum sind Funktionen nützlich?

- **Wiederverwendung:** einmal schreiben, überall aufrufen.
- **Eine Stelle zum Korrigieren:** Fehler wird an genau einer Stelle behoben, nicht an zwölf.
- **Lesbarkeit:** `flaeche_kreis(r)` sagt, *was* passiert; die Formel ist Detail.
- **Testbarkeit:** eine Funktion mit klarem Ein- und Ausgang kann man gezielt prüfen.
- **Kapselung:** Variablen innerhalb einer Funktion sind lokal und stören den Rest nicht.

## 5. Schleifen

Eine **Schleife** wiederholt einen Codeblock. Zwei Varianten:

- **`for`** — wenn die Anzahl der Durchläufe *vorher feststeht* (über eine Sammlung laufen).
- **`while`** — wenn wiederholt wird, *solange eine Bedingung gilt*.

Auch hier gilt: Doppelpunkt und Einrückung.

### 5.1 `for`-Schleife

````python
# ueber eine Liste laufen
for w in [10, 20, 30]:
    print(w)
````

````python
# ueber einen Zahlenbereich laufen: range(start, stop)  -- stop nicht enthalten
for i in range(5):
    print(i, i**2)
````

````python
# Index und Wert gleichzeitig
messwerte = [2.5, 3.1, 4.7]
for i, w in enumerate(messwerte):
    print(f"Eintrag {i}: {w}")
````

### 5.2 `while`-Schleife

````python
# laeuft, solange die Bedingung wahr ist
h = 1.0
schritt = 0
while h > 0.1:
    h = h / 2
    schritt = schritt + 1
    print(f"Schritt {schritt}: h = {h}")
````

````python
# break beendet die Schleife sofort, continue springt zum naechsten Durchlauf
for i in range(10):
    if i == 3:
        continue      # 3 ueberspringen
    if i > 6:
        break         # ab 7 abbrechen
    print(i)
````

**Achtung Endlosschleife:** Wenn die Bedingung eines `while` nie falsch wird, läuft die Zelle ewig (`[*]` bleibt stehen). Dann: *Kernel → Interrupt* bzw. der `■`-Knopf.

**Und ein Hinweis für später:** Für numerische Rechnungen sind Schleifen über NumPy-Arrays langsam. `a * 2` ist dasselbe wie eine Schleife über alle Elemente — aber um Größenordnungen schneller.
